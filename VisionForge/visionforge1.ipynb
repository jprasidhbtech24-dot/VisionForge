```python
import requests
import base64
from IPython.display import display, Image

url = "http://127.0.0.1:7860/generate"

payload = {
    "prompt": "A futuristic city in India at sunset, cinematic lighting, ultra detailed",
    "mode": "fast",
    "seed": 42,
    "classical_enhance": False
}

response = requests.post(url, json=payload, timeout=300)

print("Status:", response.status_code)

if response.status_code == 200:
    data = response.json()
    img = base64.b64decode(data["image"])
    display(Image(data=img))
    print("Generation time:", data.get("generation_ms"), "ms")
    print("CLIP score:", data.get("clip_score"))
else:
    print(response.text)
```

```python
import requests
import base64

response = requests.post(
    "http://127.0.0.1:7860/generate",
    json={
        "prompt": "A futuristic city in India at sunset, cinematic lighting, ultra detailed",
        "mode": "fast",
        "seed": 42,
        "classical_enhance": False
    },
    timeout=300
)

data = response.json()

with open("/content/visionforge_first_image.png", "wb") as f:
    f.write(base64.b64decode(data["image"]))

print("Image saved successfully!")
print("Generation time:", data["generation_ms"], "ms")
print("CLIP score:", data["clip_score"])
```

```python
import requests
import pandas as pd
import time

modes = ["fast", "balanced", "quality"]

results = []

prompt = "A futuristic city in India at sunset, cinematic lighting, ultra detailed"

for mode in modes:
    print("Testing:", mode)

    for run in range(3):
        response = requests.post(
            "http://127.0.0.1:7860/generate",
            json={
                "prompt": prompt,
                "mode": mode,
                "seed": 42,
                "classical_enhance": False
            },
            timeout=300
        )

        response.raise_for_status()
        data = response.json()

        results.append({
            "Mode": mode,
            "Run": run + 1,
            "Generation_ms": data["generation_ms"],
            "CLIP_score": data["clip_score"],
            "GPU_memory_MB": data.get("peak_gpu_memory_mb")
        })

df = pd.DataFrame(results)

display(df)

df.to_csv("visionforge_benchmark.csv", index=False)

print("Benchmark saved successfully!")
```

```python
summary = df.groupby("Mode").agg(
    Average_Time_ms=("Generation_ms", "mean"),
    Std_Dev_ms=("Generation_ms", "std"),
    Average_CLIP=("CLIP_score", "mean"),
    GPU_Memory_MB=("GPU_memory_MB", "mean")
).round(4)

display(summary)

summary.to_csv("VisionForge_Final_Evaluation.csv")

print("Evaluation summary generated!")
```
