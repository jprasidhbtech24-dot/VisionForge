## Start the application
Run this cell, then open the public Gradio-style Colab proxy URL printed by the next cell. If the proxy is unavailable, use the Hugging Face Docker Space deployment described in README.md.

```
import subprocess, time, os
log = open('/tmp/visionforge.log','w')
proc = subprocess.Popen(['uvicorn','app.main:app','--host','0.0.0.0','--port','7860'], stdout=log, stderr=log)
time.sleep(8)
print('Server PID:', proc.pid)
print(open('/tmp/visionforge.log').read()[-2000:])
```
